```python
import pandas as pd

# Load raw dataset
df = pd.read_csv("/content/drive/MyDrive/SoftCom Project/Dataset/Softcomp Annotation.csv")

# Rename first column
df = df.rename(columns={
    df.columns[0]: "image_id"
})

# Remove leading/trailing whitespace from text fields
for col in ["image_id", "wall_type",
            "description_en", "description_bn",
            "color_idea"]:
    df[col] = df[col].astype(str).str.strip()

# Standardize wall_type whitespace
df["wall_type"] = df["wall_type"].str.replace(
    r"\s+", " ", regex=True
)

# Standardize indoor values
df["indoor"] = pd.to_numeric(df["indoor"], errors="coerce")

# Standardize color formatting
df["color_idea"] = (
    df["color_idea"]
    .str.lower()
    .str.split(",")
    .apply(lambda x: ",".join(
        sorted(set(i.strip() for i in x))
    ))
)

# Check result
print(df.head())
print(df.shape)

print("\nWall Type:")
print(df["wall_type"].value_counts())

print("\nIndoor:")
print(df["indoor"].value_counts())
```

##Quality check

```python
print("Total rows:", len(df))

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate image IDs:")
print(df["image_id"].duplicated().sum())

print("\nUnique wall types:")
print(df["wall_type"].unique())

print("\nIndoor values:")
print(df["indoor"].unique())
```
